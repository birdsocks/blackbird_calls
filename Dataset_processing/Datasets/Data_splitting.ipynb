```python
import pandas as pd
import os

# === CONFIGURATION ===
csv_path = '/home/Shelby/blackbird_calls/Experiments/Detection_of_calls/Datasets/All_call_classes/Split_data_louis/cv4e_calls_channel1_splitrandom.csv'
images_dir = '/home/Shelby/blackbird_calls/Experiments/Detection_of_calls/Datasets/All_call_classes/YOLO_v1/images'
labels_base_dir = '/home/Shelby/blackbird_calls/Experiments/Detection_of_calls/Datasets/All_call_classes/YOLO_v1/labels'

# Image and spectrogram parameters (from Data_processing_imagegenerator.ipynb)
img_w_px, img_h_px = 1280, 720
segment_length = 5  # seconds per image
freq_min, freq_max = 0, 24000  # Hz

# Class mapping from blackbird.yaml
class_mapping = {
    'Check': 0,
    'Cheer': 1,
    'Chits': 2,
    'Growl': 3,
    'A': 4,
    'B': 5,
    'C': 6,
    'D': 7,
    'E': 8,
    'F': 9,
    'G': 10,
    'H': 11,
    'I': 12,
    'J': 13,
    'K': 14,
    'O': 15
}

# Load annotations
df = pd.read_csv(csv_path)

# Create output directory
os.makedirs(labels_base_dir, exist_ok=True)

# Get all image files
if not os.path.exists(images_dir):
    print(f"ERROR: Images directory not found: {images_dir}")
else:
    image_files = [f for f in os.listdir(images_dir) if f.endswith('.png')]
    print(f"Found {len(image_files)} images to process")
    
    # Process each image
    labels_created = 0
    annotations_written = 0
    skipped_annotations = 0
    
    for img_name in image_files:
        # Parse image filename: e.g., "AZ02_Trial_1_trim_spec_00-05.png"
        if '_spec_' not in img_name:
            continue
            
        # Extract file_id and time segment
        stem = img_name.replace('.png', '')
        parts = stem.split('_spec_')
        file_id = parts[0]  # e.g., "AZ02_Trial_1_trim"
        
        try:
            time_part = parts[1]  # e.g., "00-05"
            seg_start, seg_end = [int(x) for x in time_part.split('-')]
        except:
            print(f"Warning: Could not parse time segment from {img_name}")
            continue
        
        # Find annotations that overlap with this time segment
        annotations = df[
            (df['file_id'] == file_id) &
            (df['begin_time_s'] < seg_end) &
            (df['end_time_s'] > seg_start)
        ]
        
        # Create label file
        label_filename = stem + '.txt'
        label_path = os.path.join(labels_base_dir, label_filename)
        
        with open(label_path, 'w') as f:
            for idx, row in annotations.iterrows():
                # Get class number
                call_type = row['call_type']
                if call_type not in class_mapping:
                    print(f"Warning: Unknown call_type '{call_type}' in {img_name}")
                    skipped_annotations += 1
                    continue
                
                class_num = class_mapping[call_type]
                
                # Clip annotation to segment bounds
                call_start = max(row['begin_time_s'], seg_start)
                call_end = min(row['end_time_s'], seg_end)
                call_low = max(row['low_freq_hz'], freq_min)
                call_high = min(row['high_freq_hz'], freq_max)
                
                # Convert to pixel coordinates (time and frequency)
                # X: time coordinate (left to right)
                # Y: frequency coordinate (bottom to top, but image coordinates are top to bottom)
                
                # Time coordinates (normalized 0-1 within segment)
                x_left = (call_start - seg_start) / segment_length
                x_right = (call_end - seg_start) / segment_length
                
                # Frequency coordinates (normalized 0-1)
                # In spectrogram: y=0 is top (high freq), y=1 is bottom (low freq)
                # So we need to invert the frequency
                y_bottom = (call_low - freq_min) / (freq_max - freq_min)  # normalized freq
                y_top = (call_high - freq_min) / (freq_max - freq_min)  # normalized freq
                
                # In image coordinates (0 at top), high frequency is at top
                # So we invert: image_y = 1 - freq_normalized
                y_image_top = 1 - y_top
                y_image_bottom = 1 - y_bottom
                
                # Calculate YOLO format (x_center, y_center, width, height) - all normalized
                x_center = (x_left + x_right) / 2
                y_center = (y_image_top + y_image_bottom) / 2
                width = x_right - x_left
                height = y_image_bottom - y_image_top
                
                # Ensure values are within [0, 1]
                x_center = max(0, min(1, x_center))
                y_center = max(0, min(1, y_center))
                width = max(0, min(1, width))
                height = max(0, min(1, height))
                
                # Write YOLO format line: class x_center y_center width height
                f.write(f"{class_num} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}\n")
                annotations_written += 1
        
        labels_created += 1
    
    # Print summary
    print(f"\n=== Label Generation Complete ===")
    print(f"Total label files created: {labels_created}")
    print(f"Total annotations written: {annotations_written}")
    print(f"Skipped annotations (unknown call_type): {skipped_annotations}")
    print(f"\nLabel files saved to: {labels_base_dir}")
```
